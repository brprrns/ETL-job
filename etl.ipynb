**2. Deduce the percentage of developers who knew python in each country.**

```python
df.columns
```

```python
df[['Country', 'LanguageWorkedWith']].head()
```

```python
q2_df = df[["Country", "LanguageWorkedWith"]].dropna(subset = ["Country"])
```

```python
q2_df
```

```python
# Developers who knows python
q2_df["KnowsPython"] = (
    q2_df["LanguageWorkedWith"]
    .fillna("")
    .str.split(";")
    .apply(lambda languages: "Python" in languages)
)
```

```python
#Calculate persentage
q2_report = (
    q2_df.groupby("Country")["KnowsPython"]
    .mean()
    .mul(100)
    .reset_index(name="PythonPercentage")
)
```

```python
q2_report
```

**5. Generate the report for the people who code as a hobby based on their gender
and continent.**

```python
df["Gender"].value_counts(dropna=False).head(20)
```

```python
def clean_gender(gender):
    if pd.isna(gender):
        return "OTHERS"

    gender = gender.strip().upper()

    if gender == "MAN":
        return "MAN"
    elif gender == "WOMAN":
        return "WOMAN"
    else:
        return "OTHERS"
```

```python
df["GenderCategory"] = df["Gender"].apply(clean_gender)
```

```python
df["GenderCategory"].value_counts()
```

```python
q5_df = df[
    ["Country", "Hobbyist", "GenderCategory"]
].copy()

q5_df["Continent"] = coco.convert(
    q5_df["Country"],
    to="continent"
)
```

```python
q5_df["Hobbyist"].value_counts(dropna=False)
```

```python
q5_hobbyists = q5_df[
    q5_df["Hobbyist"] == "Yes"
].copy()
```

```python
q5_report = (
    q5_hobbyists
    .dropna(subset=["Continent"])
    .groupby(["Continent", "GenderCategory"])
    .size()
    .reset_index(name="HobbyistCount")
)
```

```python
q5_report
```

****6. Generate the report for job and career satisfaction of developers based on their
gender and continent.****

```python
q6_df = df[
    ["Country", "GenderCategory", "JobSat", "CareerSat"]
].copy()
```

```python
q6_df["Continent"] = coco.convert(
    q6_df["Country"],
    to="continent"
)
```

```python
q6_df["JobSat"].value_counts(dropna=False)
```

```python
q6_df["CareerSat"].value_counts(dropna=False)
```

```python
job_satisfaction_report = (
    q6_df
    .dropna(subset=["Continent", "JobSat"])
    .groupby(
        ["Continent", "GenderCategory", "JobSat"]
    )
    .size()
    .reset_index(name="DeveloperCount")
)
```
